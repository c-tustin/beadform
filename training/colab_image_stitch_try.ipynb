# Try the saved image adapter
Use a free GPU and the same Google Drive account as training. Run the two cells below. **This notebook never trains or changes model weights.** It loads the completed, audited image-stitch-v1 adapter already saved in your Drive.

Click **Use sample triangle**, then **Generate pattern**, or upload one small flat reference and provide bead dimensions. Experimental: synthetic challenge exact recipes improved from 12/40 to 26/40, while unsupported-input rejection worsened from 9/10 to 6/10. Starting/finishing and physical beadability are unverified.

[Read the complete results and failures](https://github.com/c-tustin/beadform/blob/main/training/results/image-stitch-v1/README.md). Uploads remain in your own Colab/Drive; they are not automatically published or trained on.

```
import sys,subprocess,shutil,runpy
from pathlib import Path
import torch
if not torch.cuda.is_available():
    raise RuntimeError('Connect an eligible free GPU runtime. Do not upgrade or bypass quotas.')
from google.colab import drive
drive.mount('/content/drive',timeout_ms=600000)
repo=Path('/content/beadform-image-inference-v1')
revision='06139f4deace40355c705ec7767f2c2a769c22aa'
if not repo.exists():
    subprocess.run(['git','clone','https://github.com/c-tustin/beadform.git',str(repo)],check=True)
if subprocess.check_output(['git','status','--porcelain','--untracked-files=no'],cwd=repo,text=True).strip():
    raise RuntimeError('Preserve source edits before loading this pinned interface.')
subprocess.run(['git','fetch','origin',revision],cwd=repo,check=True)
subprocess.run(['git','checkout','--detach',revision],cwd=repo,check=True)
if not shutil.which('node'):
    subprocess.run(['apt-get','update','-qq'],check=True)
    subprocess.run(['apt-get','install','-y','-qq','nodejs'],check=True)
subprocess.run([sys.executable,'-m','pip','install','-q','-r',str(repo/'training/requirements-catalog-v2.txt')],check=True)
print('GPU:',torch.cuda.get_device_name(0),'Interface source:',revision)
print('Setup only: no training or weight changes.')
```
