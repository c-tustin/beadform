# Try your trained Beadform model
Use a free GPU, connect Drive, then run the two cells. No retraining or paid endpoints. Optional image input is experimental: the adapter was trained on text only. Supply bead dimensions in text. A resized uploaded reference is saved with results in your own Drive, never automatically published.

```
import sys,subprocess,shutil
from pathlib import Path
import torch
if not torch.cuda.is_available():
    raise RuntimeError('Connect an eligible free GPU runtime. No paid upgrade or quota bypass.')
from google.colab import drive
drive.mount('/content/drive',timeout_ms=600000)
trial_repo=Path('/content/beadform-stitch-try')
if not trial_repo.exists():
    subprocess.run(['git','clone','--branch','main','https://github.com/c-tustin/beadform.git',str(trial_repo)],check=True)
if not (trial_repo/'training/colab_stitch_try.py').exists():
    raise RuntimeError('This checkout predates the image UI. Preserve edits and use a fresh inference checkout.')
print('Inference source:',subprocess.check_output(['git','rev-parse','HEAD'],cwd=trial_repo,text=True).strip())
if not shutil.which('node'):
    subprocess.run(['apt-get','update','-qq'],check=True)
    subprocess.run(['apt-get','install','-y','-qq','nodejs'],check=True)
subprocess.run([sys.executable,'-m','pip','install','-q','-r',str(trial_repo/'training/requirements-catalog-v2.txt')],check=True)
```
